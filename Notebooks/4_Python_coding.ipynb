## Importing modules <a id="Importing"></a>


A module (`modulename`) can be loaded by using the `import` command:


```
import modulename
```

To import a function(`functionname`) inside a module(`modulename`):

```
modulename.functionname()
```

A module can be imported with an **alias**:

```
import modulename as x
x.functionname()
```

Specific components of a module can be imported:

```
from x import y
```

or:


```
from x import *
```

Aliases can also be used with selective imports: 

```
from x import y as z
```

To reload a module:

```
reload(modulename)

```

### Loops <a id="Loops"></a>

The most common loop is the `for` loop:

```
for var in set:
    commands
```

There is one very useful command that produces a list output that can be used in the loop:

````python
range(4)
````

It can also take 2 or 3 arguments in the form:

````python
range(5,-3,-2)
````

We provide a simple example in which the `for` loop is used to evaluate the **sum** of the first 100 integer numbers:

````python
S = 0

for k in range(101):
    S = S+k

print(S)
````

The second type of loop is the `while` loop:

```
while condition:
    commands
```

We provide a simple example in which the `while` loop is used to evaluate again the **sum** of the first 100 integer numbers:

````python
S = 0
k = 1

while k < 101:
    S = S+k
    k = k+1

print(S)
````

Next code show the use of `break` command:

````python
S = 0
k = 1

while 1:
    S = S+k
    k = k+1
    if k > 100:
        print('End!')
        break

print(S)
````

Next code show the use of `continue` command to **exclude** 50 from the sum:

````python
S = 0
k = 1

while k < 101:
    if k == 50:
        k = k+1
        continue
    S = S+k
    k = k+1

print(S)
````

## Iterators and comprehensions <a id="Iterators"></a>

Often we need to iterate not only the values in an array, but also keep track of the index. This could be done as:

````python
L = [2, 4, 6, 8]

for i in range(len(L)):
    print(i, L[i])
````

However, Python provides a **clearer** syntax:


````python
for i, val in enumerate(L):
    print(i, val)
````

To iterate over **multiple** sequences simultaneously:

````python
L = [2, 4, 6, 8, 10]
R = [3, 6, 9, 12, 15]

for lval, rval in zip(L, R):
    print(lval, rval)
````

**List comprehension** provide a concise and readable way to build lists. Instead of:

````python
L = []

for n in range(10):
    L.append(n ** 2)

print(L)
````

we can simply write the list comprehension:


````python
[n ** 2 for n in range(10)]
````

For multiple variables:

````python
[(i, j) for i in range(2) for j in range(3)]
````

The iteration can be further controlled by adding a **conditional** expression:

````python
[val for val in range(20) if val % 3 > 0]
````

or:

````python
[val if val % 2 else -val for val in range(20) if val % 3]
````

We can easily produce a dictionary with comprehension:

````python
{n: n**2 for n in range(6)}
````

Similarly, comprehension can be used to generate a set:

````python
{n**2 for n in range(12)}
````

Recall that a set contains only **unique** elements:

````python
{a % 3 for a in range(1000)}
````

A **generator** expression is essentially a list comprehension in which elements are generated *lazily*:

````python
(n**2 for n in range(12))
````

This does not immediately produce the values, but rather an *iterable object*:

````python
G = (n**2 for n in range(12))
list(G)
````

However, generators can only be iterated **once**:

````python
G = (n**2 for n in range(12))
list(G)
````

````python
list(G)
````

This can be very **useful** because it means iteration can be stopped and started:

````python
G = (n**2 for n in range(12))

for n in G:
    print(n, end=' ')
    if n > 30: break

print("\nDoing something in between ...")

for n in G:
    print(n, end=' ')
````

## Functions <a id="Functions"></a>


A new function can be **defined** with the following syntax:

```
def funname(args):
    commands
    
    return value
```

As an example, it is provided a function that computes the hypotenuse of a triangle given the other two legs ($a$ and $b$):

````python
def pythagoras(a, b):
    """ Computes the hypotenuse of two arguments """
    c = (a**2 + b**2) ** 0.5   # 0.5 is the square root
    
    return c
````

When you make functions you can allow for **default** values: 

```
def pythagoras(a=3, b=4):
    ...
```

Functions can be naturally extended when *more than one* output variables:

````python
def rectangle(w, h=10):
    """ Computes the surface and perimeter given two sides """
    s = w * h
    p = 2*(w + h)
    
    return s, p
````

which will be used as:

````python
width = 20
height = 8
A, P = rectangle(width, height)
print("Surface =", A, "-- Perimeter =", P)
````

**Global** variables must be explicitly declared:

```
def myfun(x, y, ..):
    global A   # A is a global variable
    ...
```

Caution is required when working with **mutable** objects (*side effect*):

````python
def myfun(n, list1, list2):
    list1.append(3)  # Modifies the list1
    list2 = [7, 8, 9]
    n = n + 1
				
x, y, z = 5, [1, 2], [4, 5]
myfun(x, y, z)
x, y, z
````

When a function parameter is preceded by a single asterisk (*), it collects additional positional arguments into a tuple:

````python
def myfun(a, b, *c):
    S = (a + sum(list(c)))/b   
    return S
				
myfun(2, 1, 4, 5, 6)
````

Similarly, using a double asterisk (**) allows the function to accept an arbitrary number of keyword arguments, which are stored in a dictionary:

````python
def myfun2(x, y, **other):
    print("x: {0}, y: {1}, keys in 'other': {2}".format(x, y, list(other.keys())))
				
myfun2(2, 1, foo=4, bar=5)
````

### The decorator

Python **decorators** allow functions to be *wrapped* by other functions:

````python
def decorator(func):
    def wrapper():
        print("Before.")
        func()
        print("After.")
    return wrapper

@decorator
def say_hello():
    print("Hello!")
````

This is a *syntactic sugar*:

```
say_hello = decorator(say_hello)
say_hello()
```

Using directly the **decorated** function `say_hello()` will produce:

````python
say_hello()
````

**Anonymous** functions can be defined as:

```
func = lambda args : command
```

For example:

````python
f = lambda x : x**3 + 7 
f(7)
````

Another example:


````python
add = lambda x, y: x + y
add(1, 2)
````

Python the `map` utility:

```
newlist = map(function, list)
```

For example:

````python
L = [3, 2, 4, 1]
fun = lambda x : x**3 + 7
L2 = map(fun, L)
print(*L2)    # or list(L2)
````

Another utility is the `filter` command:

```
newlist = filter(condition, list)
```

For example:

````python
L = [3, 2, 4, 1]
cond = lambda x : x>=3
L2 = filter(cond, L)
print(*L2)    # or list(L2)
````

A **generator function** produces a sequence of values: 

````python
def gen():
    for n in range(12):
	    yield n ** 2
				
G = gen()
print(*G)
````

## Classes and objects <a id="Classes"></a>

Python is fully *object-oriented*, and **classes** are defined as:

```
class myclass(superclass):

    def __init__(self, args):

    def functionname(self, args):
```

In a class's method, the **first** argument is usually called `self` by convention:

````python
class Complex(object):
    """Represents a complex number."""

    def __init__(self, re, im):
        self.re = re
        self.im = im

    def __str__(self):
        return "%d%+dj" % (self.re, self.im)
						
    def real(self):
        return self.re
  
    def imag(self):
        return self.im					

    def abs(self):
        """Return the magnitude."""
        return (self.re**2 + self.im**2) ** 0.5
    

    # Operation overloading
    def __add__(self, other):
        """Returns the sum of the numbers."""
        new_re = self.re + other.re
        new_im = self.im + other.im
        return Complex(new_re, new_im)
    
    def __mul__(self, other):
        """Returns the product of the numbers."""
        new_re = self.re * other.re - self.im * other.im
        new_im = self.re * other.im + self.im * other.re 
        return Complex(new_re, new_im)
    
    
    def __lt__(self, other):
        """Returns |self| < |other|."""
        m1 = self.abs()
        m2 = other.abs()
        return m1 < m2
    
    def __gt__(self, other):
        """Returns |self| > |other|."""
        m1 = self.abs()
        m2 = other.abs()
        return m1 > m2   
    
    def __eq__(self, other):
        """Returns self == other."""
        return ((self.re == other.re) and (self.im == other.im))
````

Using the previous class:

````python
z1 = Complex(3, 5)
z2 = Complex(2, -1)
print(z1.real())
````

````python
z = z1 + z2
print(z)
````

````python
print(z1 < z2)
````

In order to **get** the value of a variable or to change its current value, we can resort to the **getter** and **setter** methods:

````python
class People():
    def __init__(self, name, surname, age=0):
        self.name = name
        self.surname = surname
        self.__age = age
        
    # Getter method
    def get_age(self):
        return self.__age
    
    # Setter method
    def set_age(self, a):
        self.__age = a
````

Alternatively, it is possible to create a **property** from the getter and setter :

```
age = property(get_age, set_age)
```

Otherwise, to reach the same purpose we can **decorate** the getter and the setter:

```
    # Getter method
	@property
    def age(self):
        return self.__age
    
    # Setter method
	@age.setter
    def age(self, a):
        self.__age = a
```

**Static** methods are defined by using the decorator `@staticmethod`:

````python
class MyClass():
    def __init__(self, value):
        self.value = value

    @staticmethod
    def get_max_value(x, y):
        return max(x, y)


obj = MyClass(10)  # Create an instance of MyClass
print(MyClass.get_max_value(20, 30))  
print(obj.get_max_value(20, 30))
````

**Class** methods are defined by using the decorator `@classmethod`:

````python
from datetime import date

class Person():
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod   # used to create the object by birth year
    def fromBirthYear(cls, name, year):
        return cls(name, date.today().year - year)

person = Person.fromBirthYear('Michele', 2000)
print(person.age)
````

**Inheritance** allows a class (*child* or *derived* class) to reuse and extend the functionality of another class (*parent* or *base* class):

````python
class Person():   # Parent or base class
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname  = lname

    def printname(self):
        print(self.firstname, self.lastname)

class Student(Person):   # Child or derived class
    def __init__(self, fname, lname, year):
        super().__init__(fname, lname)
        self.graduationyear = year

    def welcome(self):
        print("Welcome", self.firstname, self.lastname, "to the class of", self.graduationyear)
						
person1 = Person('Michele', 'Scarpiniti')
person2 = Student('Mario', 'Rossi', 2020)

person1.printname()

person2.printname()
		
person2.welcome()
````
