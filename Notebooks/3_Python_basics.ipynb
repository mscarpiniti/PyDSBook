### Lists <a id="Lists"></a>

A **list** is a collection of elements, enclosed in square brackets:

```python
L1 = [0, 3, 2, 'hi']
```

Lists of lists:

```python
L2 = [3, 2, [5, 4, 3], [2, 3, 2]]
```

Elements are accessed by index. **Python**  indices start at **0** :

```python
L2[0]
```

```python
L2[1]
```

However, we have also:

```python
L2[2]
```

```python
L2[3]
```

**Nested** elements require multiple indices:

**Negative** indices allow access from the end:

```python
L2[-1]
```

A useful feature is the *slice* operator:

```python
L2[2:4]
```

The *slice* operator can include a step:

```python
L2[0:4:2]
```

```python
L2[::-1]
```

```python
L1 = [None] * 4
print(L1)
```

```python
L3 = [2] * 5
print(L3)
```

The **length** of a list is evaluated using the `len` command:

```python
len(L2)
```

```python
L = [3, 2, 4, 1]
print(*L)
```

Also for lists, the **+** operator produces the **concatenation**:

```python
L1 = [3, 2, 4, 1]
L2 = [-2, 0, 5]

L = L1 + L2
print(L)
```

Elements can be **modified** via *assignment*:

We provide other examples:

```python
L.index(0)
```

Because Python is object-oriented, all variable names are simply **references** to objects:

```python
L1 = [3, 2, 4, 1]
L2 = L1
L2[2] = 10
print(L1)
```

The slice operator can **partially** solve this issue:

```python
L1 = [3, 2, 4, 1]
L2 = L1[:]
L2[2] = 10
print(L1)
```

However, the previous trick **does not**  work properly in the case of lists of lists:

```python
L1 = [3, 2, [5, 4, 3], [2, 3, 2]]
L2 = L1[:]
L2[2][1] = 10
print(L1)
```

The solution consists in performing a **deep copy** (instead of a **shallow copy**):

```python
import copy

L1 = [3, 2, [5, 4, 3], [2, 3, 2]]
L2 = copy.deepcopy(L1)
L2[2][1] = 10
print(L1)
```

### Sets <a id="Sets"></a>

A **set** is an unordered collections of **unique** items defined by curly brackets:

```python
primes = {2, 3, 5, 7}
odds = {1, 3, 5, 7, 9}
```

Common operations include **union** , **intersection**, **difference**, and **symmetric difference**:

```python
U = primes | odds    # Union
print(U)
```

```python
I = primes & odds    # Intersection
print(I)
```

```python
D1 = primes - odds   # Difference (in primes but not in odds)
print(D1)
```

```python
D2 = primes ^ odds   # Symmetric Difference (only in one set but not in both)
print(D2)
```

However, equivalent methods are also available:

```python
U = primes.union(odds)         # Union
print(U)
```

```python
I = primes.intersection(odds)  # Intersection
print(I)
```

```python
D = primes.difference(odds)    # Difference
print(D)
```

The number of element in a set is always obtained by the `len()` function:


```python
len(primes)
```

```python
len(odds)
```

A set can be **initialized** by a list or a tuple:

```python
S = set([1, 2, 3, 1, 3, 5])
```

Then it is possible to **add**  or **remove** items:


```python
S.add(4)
print(S)
```

```python
S.remove(5)
print(S)
```

Finally, we can **check** if an element is **inside** a set:


```python
1 in S                   
```

```python
5 in S
```

### Files <a id="Files"></a>

Files are **opened** by using the `open` function:

```python
file = open('./data/myfile.txt', 'a')   # Open a file
file.write("Hello world!\n")             # Write something
file.close()                             # Close the file
```

In order to avoid to **close** manually the file, we can resort to a  **context manager** interface:

```python
with open('./data/myfile.txt', 'a') as file:
    file.write("Hello world!\n")
```

```python
with open('./data/myfile.txt', 'r') as file:
    line = file.readline()
    print(line)
```

### String formatting  <a id="String_formatting"></a>

The `format` method considers using some brackets **marks** `{}` with positional numbers inside:

```python
"A {0} triangle is composed of {1} equal {2}".format("isosceles", "two", "sides")
```

To use variables inside:

```python
w, h = 10, 5
S = w*h
"The surface of a rectangle of sides {0} and {1} is {2}".format(w, h, S)
```

It is also possible to assign some **names** to the marks:

```python
"The surface of a rectangle of sides {width} and {height} is {surface}".format(width=w, height=h, surface=S)
```

The number *n* of characters allocated for printing can be **controlled** by:

```python
"The surface is {0:>7}".format(S)
```

The **modulo** operator (`%`) is:

```python
"A %s triangle is composed of %s equal %s" % ("isosceles", "two", "sides")
```

Similarly, if we need to print **float** number:

```python
"Surface is %f" % (5.65)
```

We can **control**  the number of total digits and decimal digits by setting the **width** (`w`) and **precision** (`p`) fields:

```python
"Surface is %6.2f" % (5.6534)
```

```python
"Surface is %-6.2f" % (5.6534)
```

The **modulo** operator can be also used for printing variables inside a **dictionary**:

```python
numbers = {'e': 2.718, 'pi': 3.14159}
print("%(pi).2f -- %(pi).4f -- %(e).2f" % numbers)
```

The third and last approach is based on the string interpolation, or simply **f-strings**:

```python
value = 42
message = f"The answer is simply {value}"
print(message)
```

Also in this case we can use the width and precision fields:

```python
pi = 3.14159
print(f"pi is {pi:{7}.{2}f}")
```
